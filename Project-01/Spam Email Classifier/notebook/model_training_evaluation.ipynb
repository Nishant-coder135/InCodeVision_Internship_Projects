```python
import pandas as pd
import joblib
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import classification_report
```

```python
# Ingesting Data
X_train=pd.read_csv("../data/X_train.csv").squeeze()
X_test=pd.read_csv("../data/X_test.csv").squeeze()
y_train=pd.read_csv("../data/y_train.csv").squeeze()
y_test=pd.read_csv("../data/y_test.csv").squeeze()
```

```python
# Ingesting the fitted text vectorizer
vectorizer=joblib.load("../model/text_vectorizer.joblib")
```

```python
# Transforming the cleaned feature vectors
X_train_tfidf=vectorizer.transform(X_train)
X_test_tfidf=vectorizer.transform(X_test)
```

```python
# Model training
model=MultinomialNB(alpha=1.0)
model.fit(X_train_tfidf, y_train)

# Model predictions
y_pred=model.predict(X_test_tfidf)

# Model evaluation
print(" !! Classification Report !!")
print(classification_report(y_test, y_pred))
```

```python
# Exporting trained model
model_payloads={
    'classifier':model
}
model_path="../model/spam_classifier.joblib"
joblib.dump(model_payloads, model_path)
print(" !! Successfully Exported Trained Model !! ")
```
