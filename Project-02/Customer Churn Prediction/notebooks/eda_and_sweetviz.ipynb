```python
import pandas as pd
import seaborn as sns
import sweetviz as sv
import json
```

```python
# Read from your central project blueprint config file
with open("../config/config.json", "r") as config_file:
    config=json.load(config_file)

# Ingesting Dataset using config
raw_data_path=config["file_paths"]["raw_data_path"]
target=config["data_schemas"]["target_variable"]
df=pd.read_csv(raw_data_path)
```

```python
# Inspecting the Data Architecture & Formats
df.info()
```

```python
# Previewing data sample cross-section
df.head().T
```

```python
# Inspecting the Class Label Distribution
sns.histplot(df[target])
```

```python
# Computing exact class label distribution rates
df[target].value_counts(normalize=True)
```

```python
# Initializing Sweetviz Profile 
report=sv.analyze(df, target_feat=target)

# Ingesting sweetviz report destination
html_path=config["file_paths"]["sweetviz_html_path"]
report.show_html(html_path, open_browser=False)
print("!! DataFrame Report Successfully created using Sweetviz !!")
```
